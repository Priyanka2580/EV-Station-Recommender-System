```python
import pandas as pd

def predict_wait_time(df, wait_model):
    """
    Uses the pre-trained wait_model (GradientBoostingClassifier) 
    to predict the probability of a queue (has_wait).
    """
    df_copy = df.copy()
    
    # Feature engineering for model input
    df_copy['service_ratio'] = df_copy['ports_out_of_service'] / df_copy['ports_total']
    df_copy['load_factor'] = df_copy['ports_occupied'] / df_copy['ports_total']
    
    # Select high-correlation features required by Model A
    features = ['ports_out_of_service', 'service_ratio', 'load_factor', 'utilization_rate', 'traffic_congestion_index']
    
    # Predict queue probability (class 1)
    df_copy['queue_probability'] = wait_model.predict_proba(df_copy[features])[:, 1]
    
    return df_copy
```
