```python
import pandas as pd

def predict_duration(df, duration_model):
    """
    Uses the pre-trained duration_model to predict average session duration.
    """
    df_copy = df.copy()
    
    # Feature engineering
    df_copy['service_ratio'] = df_copy['ports_out_of_service'] / df_copy['ports_total']
    
    # Select features required by Model C (now including is_peak_hour)
    features = ['charger_type', 'station_status', 'power_output_kw', 'utilization_rate', 'service_ratio', 'is_peak_hour']
    
    # Predict
    df_copy['predicted_duration_mins'] = duration_model.predict(df_copy[features])
    
    return df_copy
```
